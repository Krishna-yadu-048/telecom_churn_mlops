# 01 — Telecom Churn EDA
Exploratory analysis of the raw dataset before pipeline runs.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/telecom_churn.csv')
print(df.shape)
df.head()
```

```python
# Class balance
print(df['Churn'].value_counts())
print(f'Churn rate: {df["Churn"].mean():.2%}')
```

```python
# Descriptive statistics
df.describe()
```

```python
# Missing values
df.isnull().sum()
```

```python
# Feature distributions by churn label
num_cols = df.select_dtypes(include='number').columns.drop('Churn')
fig, axes = plt.subplots(3, 4, figsize=(16, 10))
for ax, col in zip(axes.flat, num_cols):
    df.groupby('Churn')[col].plot(kind='kde', ax=ax, legend=True)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Customer service calls vs churn (key churn indicator)
df.groupby(['CustServCalls', 'Churn']).size().unstack().plot(kind='bar', figsize=(10, 5))
plt.title('Customer Service Calls vs Churn')
plt.xlabel('Number of Customer Service Calls')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
